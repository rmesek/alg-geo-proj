```python
from scripts.visualizer.main import Visualizer
from scripts.helper.make_monotone import make_monotone
```

```python
polygon_example = [
    (2, 0),
    (5, 1),
    (6, 0),
    (8, 3),
    (7, 2),
    (8, 7),
    (6, 9),
    (5, 8),
    (2, 9),
    (1, 7),
    (2, 4),
    (4, 5),
    (3, 6),
    (5, 7),
    (5.5, 3),
    (2, 2),
    (1, 3),
    (0, 1),
]
```

```python
vis = Visualizer()
vis.add_polygon(polygon_example, fill=False)
vis.show()
```

```python
segments = []

D = make_monotone(polygon_example)
while D:
    edge = D.pop()
    segments.append((edge.source.as_tuple(), edge.target.as_tuple()))
```

```python
vis = Visualizer()
vis.add_line_segment(segments)
vis.show()
```
